```python
import pandas as pd
import re
import matplotlib.pyplot as plt
import sns

# File path
log_file1 = 'results_base5.log'

# Initialize empty lists to store the parsed data
depths = []
tasks = []
run_numbers = []
times = []

# Read the log file
with open(log_file1, 'r') as file:
    lines = file.readlines()

# Regular expression patterns to capture depth, tasks, run, and time
depth_task_pattern = re.compile(r"DEPTH=(\d+) TASKS=(\d+)")
run_pattern = re.compile(r"Run (\d+)")
time_pattern = re.compile(r"(\d+\.\d{6})")

# Iterate through the lines in the log file
current_depth = current_tasks = None

for line in lines:
    # Check for DEPTH and TASKS information
    match = depth_task_pattern.search(line)
    if match:
        current_depth = int(match.group(1))
        current_tasks = int(match.group(2))
    
    # Check for run number
    run_match = run_pattern.search(line)
    if run_match:
        run_number = int(run_match.group(1))
    
    # Check for time information
    time_match = time_pattern.search(line)
    if time_match:
        time_value = float(time_match.group(1))
        
        # Append data to lists
        depths.append(current_depth)
        tasks.append(current_tasks)
        run_numbers.append(run_number)
        times.append(time_value)

# Create a pandas DataFrame
df = pd.DataFrame({
    'DEPTH': depths,
    'TASKS': tasks,
    'Run': run_numbers,
    'Time': times
})

# Show the DataFrame (optional)
print(df)
```

```python
# Group the runs together and calculate mean, variance, and min
grouped = df.groupby(['DEPTH', 'TASKS'])['Time'].agg([list, 'mean', 'var', 'min']).reset_index()

# Rename the columns
grouped.columns = ['DEPTH', 'TASKS', 'Times', 'Average', 'Variance', 'Min']

# Print the grouped DataFrame
print(grouped)
```

```python
correlation = grouped[['DEPTH', 'TASKS', 'Average']].corr()
print(correlation)
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=grouped, x='DEPTH', y='Average', marker='o')
plt.title('DEPTH vs Average Time')
plt.xlabel('DEPTH')
plt.ylabel('Average Time')
plt.grid(True)
plt.show()

plt.figure(figsize=(10, 6))
sns.lineplot(data=grouped, x='TASKS', y='Average', marker='o')
plt.title('TASKS vs Average Time')
plt.xlabel('TASKS')
plt.ylabel('Average Time')
plt.grid(True)
plt.show()
```
